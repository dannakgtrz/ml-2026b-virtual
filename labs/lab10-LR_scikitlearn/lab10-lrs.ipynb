## Descripción del dataset

El dataset es sobre el consumo de combustible de los motores que crea una fábrica.

- `Model Year`, El año del motor, ej. 2014
- `Make`, el fabricante, ej. VOLVO
- `Model`, el modelo del motor, ej. S60 AWD
- `Vehicle Class`, tipo de vehículo, COMPACT
- `Engine Size`, tamaño del motor, ej. 3.0
- `Cylinders`, número de cilindros, ej. 6
- `Transmission`, modelo de la transmisión, ej. AS6
- `FUEL TYPE`, tipo de gazole, ej. Z
- `Fuel Consumption in City (L/100Km)` ej. 13.2
- `Fuel Consumption in HWY (L/100Km)` ej. 9.5
- `Fuel Consumption combined (L/100Km)` ej. 11.5
- `Fuel Consumption combined (MPG)` ej. 25
- `CO2 Emissions (g/km)` ej. 182
